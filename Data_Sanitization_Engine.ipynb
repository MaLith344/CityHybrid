#Run the Engine

```python
from core.data_inspector import DataInspector
from IPython.display import display

print("Initializing Feature-Complete Engine...")
inspector = DataInspector()

# 1. Load the data from your repo
inspector.load_csv('data/titanic.csv')

# 2. Structural Analysis Rubric Test (Shows 20 rows)
print("\n--- Generating Summary Preview (20 Rows) ---")
display(inspector.summary())

# 3. Clean and Manage Structural Irregularities
print("\n--- Executing Imputation, Outlier Handling, & Duplicate Inspections ---")
inspector.handle_missing_values(strategy='median', columns=['age'])
inspector.handle_missing_values(strategy='mode', columns=['embarked'])
inspector.remove_duplicates()
inspector.handle_outliers(columns=['fare'], action='flag')

# 4. Feature Extraction Verification
print("\n--- Scaling Features ---")
ml_matrix = inspector.merge_encoded_data(num_strategy='robust', cat_strategy='onehot')
print(f"ML Ready Shape: {ml_matrix.shape}")

# 5. Diagnostic Visualization Verification
print("\n--- Launching Subplot and Categorical Frequency Diagnostics ---")
inspector.plot_univariate_subplots(column='age')
inspector.plot_categorical_frequency(column='embarked')

# 6. Smart Multi-Type Evaluation Visualizations
print("\n--- Evaluating Smart Core Relationships & Unified Statistics ---")
inspector.plot_relationship(col1='survived', col2='age')
inspector.plot_all_associations_heatmap()
```
